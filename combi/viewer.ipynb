```python
import argparse
import os
import subprocess
from pathlib import Path
from ase.visualize import view
from readqe import read_qe_hubbard_out
from ase.io import read
```

```python
metals = [ 'TiN','VN']
adsorbate = ['Li2S6']
for metal in metals:
    for ads in adsorbate:
        out_file = f'{metal}_{ads}_combi_relax_u/{metal}_{ads}_combi_u4.out'
        if not os.path.exists(out_file):
            continue

        combi = read_qe_hubbard_out(out_file, index=':')
        print(combi.get_potential_energy())
        view(combi)
```

```python
metals = ['TiN', 'VN']
adsorbates = ['Li2S','Li2S2','Li2S4','Li2S6','Li2S8','S8']

adsorbate_path=pathlib.Path('/home/ameer_ubuntu/Git_projects/QE_2/adsorbates')
print(adsorbate_path)
slab_path=pathlib.Path('..')/'slab'/'output_files'
```

```python
from ase.build import add_adsorbate
adsorbates = ['Li2S']
metals = ['TiN','VN']
from readqe import read_qe_hubbard_out
for metal in metals:
    slab = read_qe_hubbard_out(slab_path / f'{metal}_slab_relax' / f'{metal}_atomic.out')
    for adsorbate in adsorbates:
        ads = read(adsorbate_path / f'{adsorbate}/espresso.pwo')
        atom_78_pos = slab.get_positions()[78]  # Get atom 78 (x, y, z)
        x_y_coords = atom_78_pos[:2]  # Extract just x and y
        ads.rotate(45,'z')
        combi = slab.copy()
        add_adsorbate(combi, ads, 3, position=x_y_coords, mol_index=2)
        view(combi)
        
        save_path = pathlib.Path('.')/f'{metal}_{adsorbate}_combiU_144.xyz'
        write(save_path, combi)
```

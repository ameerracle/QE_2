```python
from pathlib import Path
import sys
from ase.io import read
from ase.io import write
from ase.visualize import view
from ase.build import surface
```

```python
metals = ['TiN', 'VN']
adsorbates = ['S8']

last_frames = []
for metal in metals:
    for ads in adsorbates:
        traj_path = Path(f'./{metal}_{ads}/{metal}_{ads}_relax.traj')
        if not traj_path.exists():
            print(f'Skipping missing: {traj_path}')
            continue

        atoms_last = read(str(traj_path), index=-1)
        atoms_last.info['system'] = f'{metal}_{ads}'
        last_frames.append(atoms_last)
        print(f'Loaded last frame: {traj_path}')

if last_frames:
    view(last_frames)
else:
    print('No trajectories found.')
```

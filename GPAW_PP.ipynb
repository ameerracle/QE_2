```python
from ase.build import bulk
from gpaw import GPAW, PW, FermiDirac, Mixer

atoms = bulk('Li', 'bcc', a=3.51)

calc = GPAW(
    mode=PW(250),
    xc='PBE',
    kpts=(8, 8, 8),
    setups='paw',  # Will use Li.PBE, N.PBE, Ti.PBE from GPAW_SETUP_PATH
    occupations=FermiDirac(0.1),
    mixer=Mixer(beta=0.05, nmaxold=5, weight=50),
    txt='-',
)
atoms.calc = calc
e = atoms.get_potential_energy()
print(f"Energy: {e:.6f} eV")
```

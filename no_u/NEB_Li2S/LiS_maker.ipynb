```python
from ase.io import read, write
from ase.visualize import view
from ase.atoms import Atoms
import pathlib
metals = ['ScN','TiN','ZrN','NbN','VN']
adsorbates = ['Li2S','Li2S2','Li2S4','Li2S6','Li2S8','S8']
Li_atom = Atoms('Li')
```

```python
for metal in metals:
    combi_path = pathlib.Path('..')/'combi_144'/'final_combi'/f'{metal}_Li2S_combi_final.xyz'
    combi = read(combi_path)
    #view(combi)
    site_pos = combi.get_positions()[79]  # Reference atom position (x, y, z)
    combi.pop(146)

    li_pos = [site_pos[0], site_pos[1], site_pos[2] +2.2 ]
    combi += Atoms('Li', positions=[li_pos])

    view(combi)
    save_path = pathlib.Path('.')/f'{metal}_LiS_combi2_144.xyz'
    write(save_path, combi)
```

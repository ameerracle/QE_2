```python
from ase.io import read
from ase.visualize import view


def safe_read_qe(path):
    # Prefer tolerant parsing for QE outputs that may have incomplete restart sections.
    return read(path, index=-1, format="espresso-out", results_required=False)


metals = ["TiN", "ZrN", "NbN", "VN", "ScN"]
adsorbates = ["Li2S8"]

for metal in metals:
    for adsorbate in adsorbates:
        restart_path = f"./{metal}_{adsorbate}_combi_relax/{metal}_{adsorbate}_combi_restart.out"
        main_path = f"./{metal}_{adsorbate}_combi_relax/{metal}_{adsorbate}_combi.out"

        try:
            combi = safe_read_qe(restart_path)
            used = restart_path
        except Exception:
            combi = safe_read_qe(main_path)
            used = main_path

        print(f"Viewing: {used}")
        view(combi)
```

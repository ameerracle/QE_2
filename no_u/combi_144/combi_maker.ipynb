```python
metals = ['ScN','TiN','ZrN','NbN','VN']
adsorbates = ['Li2S','Li2S2','Li2S4','Li2S6','Li2S8','S8']

adsorbate_path=pathlib.Path('/home/ameer_ubuntu/Git_projects/QE_2/adsorbates')
print(adsorbate_path)
slab_path=pathlib.Path('..')/'slab'/'output_slab_144'
```

```python
from ase.build import add_adsorbate
adsorbates = ['Li2S']
#metals = ['ScN','TiN','ZrN','NbN','VN']
metals = ['ScN','TiN','ZrN','NbN','VN']
for metal in metals:
    slab = read(slab_path/f'{metal}_slab_relax'/f'{metal}_slab.out')
    for adsorbate in adsorbates:
        ads = read(adsorbate_path/f'{adsorbate}/espresso.pwo')
        atom_110_pos = slab.get_positions()[110]  # Get atom 25 (x, y, z)
        x_y_coords = atom_110_pos[:2]  # Extract just x and y
        ads.rotate(45,'z')
        combi = slab.copy()
        add_adsorbate(combi, ads, 3, position=x_y_coords, mol_index=2)
        view(combi)
        save_path = pathlib.Path('.')/f'{metal}_{adsorbate}_combi_144.xyz'
        write(save_path, combi)
```

```python
from ase.io import read, write
from ase.visualize import view
from ase.build import add_adsorbate
from ase.atoms import Atoms
import pathlib
metals = ['VN']
adsorbates = ['Li2S','Li2S2','Li2S4','Li2S6','Li2S8','S8']
Li_atom = Atoms('Li')
```

```python
Li_atom = Atoms('Li')
for metal in metals:

    #slab_path = pathlib.Path('..')/'slab'/'output_files'
    combi = read('/home/ameer_ubuntu/Git_projects/QE_2/u/combi/final_xyz/VN_Li2S_final.xyz')
    site_pos = combi.get_positions()[15]
    # Reference atom position (x, y, z)
    xy_pos = site_pos[:2]  # Extract x and y coordinates
    combi.pop(146)

    add_adsorbate(combi, Li_atom, -0.6, position=xy_pos)
    view(combi)
    save_path = pathlib.Path('.')/f'{metal}_LiS_combi2u_144.xyz'
    write(save_path, combi)
```

```python
from pathlib import Path
import sys

from ase.build import surface
from ase.io import write
from ase.visualize import view

repo_root = Path.cwd().resolve().parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from readqe import read_qe_hubbard_out

slab_dir = Path.cwd().resolve() / 'output_files'
slab_dir.mkdir(parents=True, exist_ok=True)

atomic_cif_root = repo_root / 'cif' / 'atomic_cif'
final_u_root = repo_root / 'cif' / 'final_U_cif'

def make_slab_from_out(out_file, miller=(1, 0, 0), layers=2, vacuum=10.0, repeat=(3, 3, 1)):
    atoms = read_qe_hubbard_out(out_file, index=-1)
    slab = surface(atoms, miller, layers=layers, vacuum=vacuum)
    if repeat is not None:
        slab = slab.repeat(repeat)
    return slab
```

```python
# Slabs from the attached atomic and orthorhombic atomic vc-relax outputs
source_groups = {
    'atomic_cif': {
        'TiN_atomic': atomic_cif_root / 'TiN_atomic' / 'TiN_atomic_vcrelax.out',
        'TiN_ortho_atomic': atomic_cif_root / 'TiN_ortho_atomic' / 'TiN_ortho_atomic_vcrelax.out',
        'VN_atomic': atomic_cif_root / 'VN_atomic' / 'VN_atomic_vcrelax.out',
        'VN_ortho_atomic': atomic_cif_root / 'VN_ortho_atomic' / 'VN_ortho_atomic_vcrelax.out',
    },
}

for group_name, sources in source_groups.items():
    for label, out_file in sources.items():
        slab = make_slab_from_out(out_file)
        view(slab)
        output_file = slab_dir / f'{label}_100_slab.xyz'
        write(output_file, slab)
        print(f'Wrote {output_file}')
```

```python
# Slabs from the final_U_cif relaxed vc-relax outputs
final_u_sources = {
    'TiN': final_u_root / 'TiN' / 'vcrelax.out',
    'VN': final_u_root / 'VN' / 'vcrelax.out',
    'ZrN': final_u_root / 'ZrN' / 'vcrelax.out',
    'ScN': final_u_root / 'ScN' / 'vcrelax.out',
    'NbN': final_u_root / 'NbN' / 'vcrelax.out',
}

for label, out_file in final_u_sources.items():
    slab = make_slab_from_out(out_file)
    output_file = slab_dir / f'final_U_cif_{label}_144_slab.xyz'
    write(output_file, slab)
    print(f'Wrote {output_file}')
```

```python
from ase.io import read,write
from ase.visualize import view

metals = ['TiN','VN','ScN','ZrN','NbN']
for metal in metals:
    super = read(f"./primitive/{metal}_prim.poscar")
    #view(cif)
    mega = super.repeat((2, 2, 2))
    view(mega)
    write(f"./supercell/{metal}_supercell.cif", mega)
```
